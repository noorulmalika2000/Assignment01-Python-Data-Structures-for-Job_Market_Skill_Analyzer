```python
#Create the initial job dataset
job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": ["Data Analyst","Data Scientist","Business Analyst","Data Engineer","ML Engineer"],
    "Company": ["ABC Corp","XyZ L","DataWorks","InfraTech","AI Labs"],
    "Skills": ["Excel SQL Python","Python ML Statistics","Excel SQL PowerBI","Python SQL Spark","Python ML DeepLearning"]
}

print(job_data)
```

```python
#Display job postings in a readable format
for i in range(len(job_data["Job_ID"])):
    print(
        f"ID: {job_data['Job_ID'][i]} | "
        f"Role: {job_data['Role'][i]} | "
        f"Company: {job_data['Company'][i]} | "
        f"Skills: {job_data['Skills'][i]}"
    )
```

```python
#Add new job postings
num_jobs = int(input("How many new job postings do you want to add? "))
for i in range(num_jobs):
    print("\nEnter details for job", i + 1)
    new_id = max(job_data["Job_ID"])
    role = input("Enter job role: ")
    company = input("Enter company name: ")
    skills = input("Enter skills (separated by spaces): ")
    job_data["Job_ID"].append(new_id)
    job_data["Role"].append(role)
    job_data["Company"].append(company)
    job_data["Skills"].append(skills)

print("\nUpdated job data:")
print(job_data)
```

```python
#Clean and normalize skills
def clean_skills(skills):
    skills = skills.lower()
    skills = skills.replace(",", " ")
    skills = " ".join(skills.split())
    return skills
for i in range(len(job_data["Skills"])):
    job_data["Skills"][i] = clean_skills(job_data["Skills"][i])
print("Cleaned skills:")
print(job_data["Skills"])
```

```python
#Skill extraction and frequency analysis
all_skills = []
for skills in job_data["Skills"]:
    skill_list = skills.split()
    for skill in skill_list:
        all_skills.append(skill)
skill_count = {}
for skill in all_skills:
    if skill in skill_count:
        skill_count[skill] += 1
    else:
        skill_count[skill] = 1
print("Skill counts:")
print(skill_count)
top_skills = sorted(skill_count.items(),key=lambda item: item[1],reverse=True)[:5]
print("\nTop 5 most frequent skills:")
for skill, count in top_skills:
    print(skill, ":", count)
```

```python
#Basic statistics and insights
total_jobs = len(job_data["Job_ID"])
unique_skills = set(all_skills)
python_jobs = [job_data["Role"][i]
    for i in range(total_jobs)
    if "python" in job_data["Skills"][i].split()]
print("Total Jobs:", total_jobs)
print("Unique skills:", len(unique_skills))
print("Jobs mentioning Python:", python_jobs)
```

```python
#Search jobs by skill
def find_jobs_by_skill(skill_query):
    matching_jobs = []
    for i in range(len(job_data["Job_ID"])):
        skills = job_data["Skills"][i].lower().split()
        if skill_query.lower() in skills:
            job = (job_data["Job_ID"][i],job_data["Role"][i],job_data["Company"][i])
            matching_jobs.append(job)
    return matching_jobs
search_skill = input("Enter a skill to search: ")
results = find_jobs_by_skill(search_skill)
print("Matching jobs:")
for job in results:
    print(job)
```

```python
#Find roles with at least one common skill
common_skills_jobs = filter(
    lambda job: any(
        skill_count.get(skill, 0) >= 2
        for skill in job[1].split()),
    zip(job_data["Role"], job_data["Skills"]))
roles_with_common_skills = []
for role, skills in common_skills_jobs:
    roles_with_common_skills.append(role)
print("Roles with at least one common skill:", roles_with_common_skills)
```

```python
#Convert the consolidated list to a set
unique_skills_sorted = sorted(set(all_skills))
print("Unique skills:")
print(unique_skills_sorted)
```

```python
import csv
try:
    # Save job details to CSV file
    with open("job_skills.csv", "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Job_ID", "Role", "Company", "Skills"])
        for i in range(len(job_data["Job_ID"])):
            writer.writerow([
                job_data["Job_ID"][i],
                job_data["Role"][i],
                job_data["Company"][i],
                job_data["Skills"][i]])
    # Save top skills to TXT file
    with open("top_skills.txt", "w") as file:
        file.write("Top Skills by Frequency\n")
        sorted_skills = sorted(skill_count.items(),key=lambda item: item[1],reverse=True)
        for skill, count in sorted_skills[:5]:
            file.write(f"{skill}: {count}\n")
except OSError as e:
    print("File error:", e)
else:
    print("Files saved successfully!")
finally:
    print("File handling completed.")
```
